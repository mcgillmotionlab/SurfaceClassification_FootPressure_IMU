```python
import pandas as pd
from datetime import datetime, timedelta
import numpy as np
import os
from scipy.signal import find_peaks
```

```python
def add_gait_count(folder_path):
    print("Workinig on", folder_path)   
    df = pd.read_csv(os.path.join(folder_path, "merged.csv"))
    x = df['angularVelocity_RightLowerLeg_x']
    y = df['angularVelocity_RightLowerLeg_y']
    z = df['angularVelocity_RightLowerLeg_z']

    # Combine acceleration data into a single magnitude vector
    gyr_magnitude = np.sqrt(x**2 + y**2 + z**2)
    df['gyrMagnitude_RightLowerLeg'] = gyr_magnitude

    # Find peaks in the acceleration magnitude
    peaks, _ = find_peaks(gyr_magnitude, height=0.7, distance=50)

    # Create a new column for gait cycle count
    gait_cycle_count = np.zeros(len(df), dtype=int)

    # Assign cycle numbers to the detected peaks
    for cycle_number, peak in enumerate(peaks, start=1):
        gait_cycle_count[peak:] = cycle_number

    # Add the gait cycle count to the DataFrame
    df['stepcount'] = gait_cycle_count

    # Group by 'step' and filter out groups with more than one unique 'walk_mode' label
    filtered_df = (
        df.groupby('stepcount')
        .filter(lambda x: x['walk_mode'].nunique() == 1)
    )

    # Reassign the `stepcount` column to be continuous and start from 1
    filtered_df['stepcount'] = pd.factorize(filtered_df['stepcount'])[0] + 1

    filtered_df.to_csv(os.path.join(folder_path, "merged_gait_count_shank.csv"), index=False)
```

```python
dataset_path = "data_set"

print("Adding gait count according to the shank-based gait detection...")
# Iterate through the directory structure
for course_folder in os.listdir(dataset_path):
    course_folder_path = os.path.join(dataset_path, course_folder)
    if os.path.isdir(course_folder_path):
        for subfolder in os.listdir(course_folder_path):
            subfolder_path = os.path.join(course_folder_path, subfolder)
            if os.path.isdir(subfolder_path):
                add_gait_count(subfolder_path)
print("Gait count added successfully")
```
