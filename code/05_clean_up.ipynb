```python
import os
```

```python
def cleanup(folder_path):
    print("Working on", folder_path)
    if not os.path.exists(folder_path):
        raise FileNotFoundError(f"The folder '{folder_path}' does not exist.")

    keep_files = {"gait_segmentation.csv", "merged.csv", "merged_gait_count_annotations.csv", "sliding_window.csv", "stat_freq_features(gait cycles).csv", "stat_freq_features(sliding window).csv"}  # Update with actual filenames you want to keep

    for file_name in os.listdir(folder_path):
        file_path = os.path.join(folder_path, file_name)

        if os.path.isfile(file_path) and file_name.endswith(".csv") and file_name not in keep_files:
            try:
                os.remove(file_path)
                print(f"Deleted: {file_name}")
            except PermissionError as e:
                print(f"Permission denied: {file_path}. {e}")
            except Exception as e:
                print(f"Error deleting file {file_path}. {e}")
```

```python
dataset_path = "data_set"

print("Cleaning up folders...")
for course_folder in os.listdir(dataset_path):
    course_folder_path = os.path.join(dataset_path, course_folder)
    if os.path.isdir(course_folder_path):
        for subfolder in os.listdir(course_folder_path):
            subfolder_path = os.path.join(course_folder_path, subfolder)
            if os.path.isdir(subfolder_path):
                cleanup(subfolder_path)
print("Folders cleaned up successfully")
```
