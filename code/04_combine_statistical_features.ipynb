```python
import os
import pandas as pd
```

```python
base_dir = 'data_set'

# Initialize an empty list to hold DataFrames
dfs = []

# Walk through all folders and subfolders in the base directory
for root, dirs, files in os.walk(base_dir):
    # Check if 'statistical_features.csv' is in the current folder
    if 'stat_freq_features_xsens_best_insoles_best(sliding window).csv' in files:
        csv_file_path = os.path.join(root, 'stat_freq_features_xsens_best_insoles_best(sliding window).csv')
        print("merging", csv_file_path)
        
        df = pd.read_csv(csv_file_path)
        dfs.append(df)

# Concatenate all the DataFrames vertically
combined_df = pd.concat(dfs, ignore_index=True)
combined_df.to_csv(os.path.join(base_dir, 'combined_stat_freq_features_xsens_best_insoles_best(sliding window).csv'), index=False)
print("merging complete")
```
