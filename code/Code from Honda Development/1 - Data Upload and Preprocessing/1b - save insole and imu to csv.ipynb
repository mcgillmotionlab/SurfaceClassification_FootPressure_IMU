```python
imu_sensor_locations = ['Pelvis', 'L5', 'L3', 'T12', 'T8', 'Neck', 'Head', 'RightShoulder', 'RightUpperArm', 'RightForeArm', 'RightHand', 'LeftShoulder', 'LeftUpperArm', 'LeftForeArm', 'LeftHand', 'RightUpperLeg', 'RightLowerLeg',
                    'RightFoot', 'RightToe', 'LeftUpperLeg', 'LeftLowerLeg', 'LeftFoot', 'LeftToe']

insole_sensor_locations = ['Arch', 'Hallux', 'Heel_L', 'Heel_R', 'Met1', 'Met3', 'Met5', 'Toes']

participant_num = [1, 2, 3, 4, 5, 7, 8, 10, 12, 13, 14, 15, 16, 17, 18, 19, 22, 23, 24, 25]
```

```python
# List of desired sensor locations
insole_sensor_locations = ['Arch', 'Hallux', 'Heel_L', 'Heel_R', 'Met1', 'Met3', 'Met5', 'Toes']
original_data_dir = "/content/drive/MyDrive/Next_Step/Honda: multi_modal_gait_database/processed_data"

# Create an empty list to store the DataFrames
insole_dfs = []

# Use glob to find files that match the pattern with insole sensor locations
for location in insole_sensor_locations:
    file_pattern = f"{original_data_dir}/*_{location}_*.csv"
    files = glob.glob(file_pattern)

    for file in files:
        try:
            # Load the CSV file
            df = pd.read_csv(file)

            # Debugging: Print file name and DataFrame shape
            print(f"Loaded: {file}, Shape: {df.shape}")

            # Append the DataFrame to the list
            insole_dfs.append(df)
        except Exception as e:
            # Handle any errors during file reading
            print(f"Error loading {file}: {e}")

# Combine all the DataFrames into one (optional)
if insole_dfs:
    combined_df_insole = pd.concat(insole_dfs, ignore_index=True)
    print(f"Combined DataFrame Shape: {combined_df_insole.shape}")
else:
    print("No matching files were loaded.")
```

```python
insole_df = combined_df_insole.copy()
insole_df = insole_df.sort_values(by=['participant_id', 'task', 'time', 'sensor_location'])
insole_df['walk_mode'].value_counts()
```

```python
insole_df['time_in_secs'] = insole_df['time'] / 1000
```

```python
# Perform the groupby operation and sum the selected columns
grouped_sum = insole_df.groupby(['participant_id', 'task', 'time'])[['Left_norm', 'Right_norm']].sum()

# Merge the result back into the original DataFrame
insole_df = insole_df.merge(grouped_sum, on=['participant_id', 'task', 'time'], suffixes=('', '_cumulative'))
```

```python

luo_sensor_locations = ['Pelvis', 'RightForeArm', 'RightUpperLeg', 'RightLowerLeg', 'LeftUpperLeg', 'LeftLowerLeg', 'RightFoot', 'LeftFoot']
original_data_dir = "/content/drive/MyDrive/Next_Step/Honda: multi_modal_gait_database/processed_data"

# Create an empty list to store the DataFrames
imu_dfs = []

# Use glob to find files that match the pattern with imu sensor locations
for location in luo_sensor_locations:
    file_pattern = f"{original_data_dir}/*_{location}_*.csv"
    files = glob.glob(file_pattern)

    for file in files:
        try:
            # Load the CSV file
            df = pd.read_csv(file)

            # Debugging: Print file name and DataFrame shape
            print(f"Loaded: {file}, Shape: {df.shape}")

            # Append the DataFrame to the list
            imu_dfs.append(df)
        except Exception as e:
            # Handle any errors during file reading
            print(f"Error loading {file}: {e}")

# Combine all the DataFrames into one (optional)
if imu_dfs:
    combined_df_imu = pd.concat(imu_dfs, ignore_index=True)
    print(f"Combined DataFrame Shape: {combined_df_imu.shape}")
else:
    print("No matching files were loaded.")
```

```python
imu_df = combined_df_imu.copy()
imu_df = imu_df.sort_values(by=['participant_id', 'task', 'time', 'sensor_location'])
imu_df['walk_mode'].value_counts()
```

```python
insole_df['walk_mode'] = insole_df['walk_mode'].replace({'pavement_up': 'stairs_up', 'pavement_down': 'stairs_down'})
imu_df['walk_mode'] = imu_df['walk_mode'].replace({'pavement_up': 'stairs_up', 'pavement_down': 'stairs_down'})
```
